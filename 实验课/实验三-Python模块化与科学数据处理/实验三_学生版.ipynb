## 六、综合实验：完整数据处理流程


```python
raw_from_module = data_utils.load_data(data_path)
raw_snapshot = raw_from_module.copy(deep=True)
quality_report = data_utils.inspect_data(raw_from_module)
cleaned_data = data_utils.clean_data(raw_from_module)
standardized_data = data_utils.standardize_features(cleaned_data)
label_summary = data_utils.summarize_by_label(cleaned_data)
unusual_sample_id = data_utils.find_most_unusual_sample(cleaned_data)

print("原始数据质量：", quality_report)
print("清洗后形状：", cleaned_data.shape)
print("最不寻常样本：", unusual_sample_id)
display(standardized_data)
display(label_summary)
```

### 公开自测

这些测试检查接口和关键性质。不要修改测试来迁就错误结果。


```python
assert quality_report == {
    "rows": 16,
    "columns": 5,
    "missing_cells": 2,
    "duplicate_rows": 1,
}
assert cleaned_data.shape == (11, 5)
assert cleaned_data.index.tolist() == list(range(11))
assert cleaned_data["sample_id"].is_unique
assert set(cleaned_data["label"]) == {"A", "B", "C"}
assert not cleaned_data.isna().any().any()
assert (cleaned_data[list(data_utils.FEATURE_COLUMNS)] > 0).all().all()

z_columns = [f"z_{column}" for column in data_utils.FEATURE_COLUMNS]
assert standardized_data.shape == (11, 8)
assert standardized_data.columns[-3:].tolist() == z_columns
assert np.allclose(standardized_data[z_columns].mean().to_numpy(), 0.0)
assert np.allclose(standardized_data[z_columns].std(ddof=0).to_numpy(), 1.0)

assert label_summary["label"].tolist() == ["A", "B", "C"]
assert label_summary["sample_count"].tolist() == [3, 5, 3]
assert unusual_sample_id == "S001"
pd.testing.assert_frame_equal(raw_from_module, raw_snapshot)
print("✅ 综合实验公开自测全部通过")
```

### 边界测试

缺少必需列时应抛出 `ValueError`；某个特征标准差为 0 时也应拒绝继续标准化。


```python
try:
    data_utils.clean_data(raw_from_module.drop(columns=["label"]))
except ValueError:
    pass
else:
    raise AssertionError("缺少必需列时应抛出 ValueError")

constant_data = cleaned_data.copy()
constant_data["feature_1"] = 1.0
try:
    data_utils.standardize_features(constant_data)
except ValueError:
    pass
else:
    raise AssertionError("特征标准差为 0 时应抛出 ValueError")

print("✅ 边界测试通过")
```
