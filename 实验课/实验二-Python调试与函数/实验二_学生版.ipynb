## 题 3：IndexError

找出列表中第一个负数的下标；没有负数时返回 `-1`。


```python
# TODO 3：修复循环边界，并保证没有负数时正确返回
def first_negative_index(values: list[int | float]) -> int:
    for index in range(len(values) + 1):
        if values[index] < 0:
            return index
```

```python
assert first_negative_index([3, 0, -2, 5]) == 2
assert first_negative_index([1, 2, 3]) == -1
assert first_negative_index([]) == -1
print("✅ 题 3 通过")
```

## 题 5：逻辑错误——条件顺序

不少于 85 分为 `excellent`，不少于 60 分为 `pass`，否则为 `fail`。


```python
# TODO 5：修复条件顺序
def score_level(score: float) -> str:
    if score >= 60:
        return "pass"
    elif score >= 85:
        return "excellent"
    else:
        return "fail"
```

```python
assert score_level(85) == "excellent"
assert score_level(84) == "pass"
assert score_level(60) == "pass"
assert score_level(59) == "fail"
print("✅ 题 5 通过")
```

## 题 6：逻辑错误——累计变量

统计列表中所有非负数的和。


```python
# TODO 6：修复累计变量的位置，并返回结果
def sum_non_negative(values: list[int | float]) -> int | float:
    for value in values:
        total = 0
        if value >= 0:
            total += value
```

```python
assert sum_non_negative([3, -2, 5, 0]) == 8
assert sum_non_negative([-3, -1]) == 0
assert sum_non_negative([]) == 0
print("✅ 题 6 通过")
```

## 题 7：print 与 return

函数应返回字典中所有杯数之和。


```python
# TODO 7：让调用者能够获得计算结果
def count_cups(category_cups: dict[str, int]) -> int:
    total = 0
    for cups in category_cups.values():
        total += cups
    print(total)
```

```python
assert count_cups({"coffee": 5, "tea": 3}) == 8
assert count_cups({}) == 0
print("✅ 题 7 通过")
```
