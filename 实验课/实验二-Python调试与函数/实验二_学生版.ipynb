## 题 1：SyntaxError

修复语法，使函数对不少于 99 元的订单免运费，否则收取 8 元。

## 题 3：IndexError

找出列表中第一个负数的下标；没有负数时返回 `-1`。

```python
# TODO 3：修复循环边界，并保证没有负数时正确返回
def first_negative_index(values):
    for index in range(len(values) + 1):
        if values[index] < 0:
            return index
```

```python
assert first_negative_index([3, 0, -2, 5]) == 2
assert first_negative_index([1, 2, 3]) == -1
assert first_negative_index([]) == -1
print("✅ 题 3 通过")
```

## 题 5：逻辑错误——条件顺序

不少于 85 分为 `excellent`，不少于 60 分为 `pass`，否则为 `fail`。

```python
# TODO 5：修复条件顺序
def score_level(score):
    if score >= 60:
        return "pass"
    elif score >= 85:
        return "excellent"
    else:
        return "fail"
```

## 题 6：逻辑错误——累计变量

统计列表中所有非负数的和。

```python
# TODO 6：修复累计变量的位置，并返回结果
def sum_non_negative(values):
    for value in values:
        total = 0
        if value >= 0:
            total += value
```

```python
assert sum_non_negative([3, -2, 5, 0]) == 8
assert sum_non_negative([-3, -1]) == 0
assert sum_non_negative([]) == 0
print("✅ 题 6 通过")
```

函数应返回字典中所有杯数之和。

```python
# TODO 7：让调用者能够获得计算结果
def count_cups(category_cups):
    total = 0
    for cups in category_cups.values():
        total += cups
    print(total)
```

## 题 9：定义实数类

修复 `RealNumber`，使它支持加法、乘法、绝对值和调试表示。运算应返回新对象，不要修改原对象。

```python
# TODO 9：修复类型转换，并让加法返回 RealNumber 对象
class RealNumber:
    def __init__(self, value):
        self.value = value

    def __repr__(self):
        return f"RealNumber({self.value})"

    def __add__(self, other):
        print(self.value + other.value)

    def __mul__(self, other):
        return RealNumber(self.value * other.value)

    def __abs__(self):
        return abs(self.value)

    def to_complex(self):
        return ComplexNumber(self.value, 0)
```

```python
x = RealNumber(3)
y = RealNumber("-2")

assert isinstance(x + y, RealNumber)
assert (x + y).value == 1.0
assert (x * y).value == -6.0
assert abs(y) == 2.0
assert repr(x) == "RealNumber(3.0)"
assert x.value == 3.0 and y.value == -2.0
print("✅ 题 9 通过")
```

## 题 10：定义并调试复数类

复数乘法满足 `(a + bi)(c + di) = (ac - bd) + (ad + bc)i`。修复乘法、模长和共轭方法。

```python
# TODO 10：修复 __mul__、__abs__ 和 conjugate
class ComplexNumber:
    def __init__(self, real, imag=0):
        self.real = float(real)
        self.imag = float(imag)

    def __repr__(self):
        return f"ComplexNumber({self.real}, {self.imag})"

    def __add__(self, other):
        return ComplexNumber(
            self.real + other.real,
            self.imag + other.imag,
        )

    def __mul__(self, other):
        return ComplexNumber(
            self.real * other.real + self.imag * other.imag,
            self.real * other.imag,
        )

    def __abs__(self):
        (self.real ** 2 + self.imag ** 2) ** 0.5

    def conjugate(self):
        self.imag = -self.imag
        return self
```

```python
z1 = ComplexNumber(3, 4)
z2 = ComplexNumber(1, -2)
z_sum = z1 + z2
z_product = z1 * z2
z_conjugate = z1.conjugate()

assert (z_sum.real, z_sum.imag) == (4.0, 2.0)
assert (z_product.real, z_product.imag) == (11.0, -2.0)
assert abs(z1) == 5.0
assert (z_conjugate.real, z_conjugate.imag) == (3.0, -4.0)
assert (z1.real, z1.imag) == (3.0, 4.0), "conjugate 不应修改原对象"
assert repr(z2) == "ComplexNumber(1.0, -2.0)"
print("✅ 题 10 通过")
```
