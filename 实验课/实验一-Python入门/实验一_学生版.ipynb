### 练习 3：循环

遍历下面的列表，分别统计正数和负数的个数，0 不计入任何一类。


```python
numbers = [3, -1, 0, 7, -5]
positive_count = 0
negative_count = 0

for item in numbers:
    if item < 0:
        negative_count += 1
    if item > 0:
        positive_count += 1

print("正数个数：", positive_count)
print("负数个数：", negative_count)
```

## 三、数据存储结构

列表适合保存有顺序、可修改的一组数据；字典用键保存一条记录的不同字段。元组适合固定数据，集合适合去重和成员判断。

### 练习 4：列表的增删改查

依次完成：在末尾添加 81；把第二个成绩改为 75；删除 60；计算平均分。


```python
scores = [88, 72, 95, 60]

# TODO：使用 append、下标赋值和 remove 修改列表
scores.append(81)
scores[1] = 75

# TODO：计算平均分并保存到 average_score
average_score = sum(scores) / len(scores)

print("修改后的成绩：", scores)
print(f"平均分：{average_score:.2f}")
```

```python
assert scores == [88, 75, 95, 81], f"列表修改错误：得到 {scores}"
assert abs(average_score - 84.75) < 1e-9, f"平均分错误：得到 {average_score}"
print("✅ 列表练习通过")
```

### 练习 5：字典的读取、更新与遍历

美式售出 2 杯；新增红茶库存 4 杯；使用 `get` 查询不存在的摩卡；遍历库存，将数量不超过 3 的饮品加入 `low_stock`。


```python
inventory = {"美式": 5, "拿铁": 3}

# TODO：更新美式库存，并添加红茶
inventory["美式"] = inventory["美式"] - 2
inventory["红茶"] = 4

# TODO：使用 get 查询摩卡，不存在时返回 0
mocha_quantity = -1
mocha_quantity = inventory.get("摩卡", 0)

low_stock = []
# TODO：使用 for 和 items() 找出库存不超过 3 的饮品
for drink, num in inventory.items():
    if num <= 3:
        low_stock.append(drink)

print("当前库存：", inventory)
print("低库存饮品：", low_stock)
```

```python
assert inventory == {"美式": 3, "拿铁": 3, "红茶": 4}, f"库存更新错误：得到 {inventory}"
assert mocha_quantity == 0, f"get 默认值错误：得到 {mocha_quantity}"
assert low_stock == ["美式", "拿铁"], f"低库存列表错误：得到 {low_stock}"
print("✅ 字典练习通过")
```
